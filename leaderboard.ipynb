# Experiment Leaderboard

Sorted by OOF ROC-AUC (descending).

```python
import pandas as pd

# Columns to display — everything else (git hashes, artifact paths, fixed CV params, etc.) is excluded
DISPLAY_COLS = [
    "rank",
    "tag",
    "model_class",
    "data_version",
    "n_features",
    "oof_roc_auc",
    # "fold_roc_auc_mean",
    # "fold_roc_auc_std",
    "oof_score",        # accuracy
    # "fold_scores_mean", # accuracy CV mean
    # "fold_scores_std",  # accuracy CV std
    "training_time_sec",
    # "status",
    "notes",
    "run_id",
]

df = pd.read_csv("experiments/runs.csv")

df = df.sort_values(["model_class", "oof_roc_auc"], ascending=False).reset_index(drop=True)
df.insert(0, "rank", df.index + 1)

leaderboard = df[DISPLAY_COLS].copy()

# Round floats for readability
float_cols = ["oof_roc_auc", "oof_score", "training_time_sec"]
leaderboard[float_cols] = leaderboard[float_cols].round(6)
leaderboard['n_features'] = leaderboard['n_features'].astype(int)

leaderboard.style.hide(axis="index")
# leaderboard[leaderboard['model_class'] == 'LGBMClassifier'].style.hide(axis="index")
```
